# Analyse des résultats

Ce notebook sert à comparer les runs après entraînement. Il ne contient pas de résultats pré-remplis.

```python
from pathlib import Path
import mlflow
import pandas as pd
import yaml

with open('../config.yaml', 'r', encoding='utf-8') as f:
    config = yaml.safe_load(f)

mlflow.set_tracking_uri('../' + config['paths']['mlflow_uri'])
experiment = mlflow.get_experiment_by_name(config['mlflow']['experiment_name'])
experiment
```

```python
if experiment is None:
    print('Aucun run MLflow pour le moment. Lancez les scripts d entraînement.')
else:
    runs = mlflow.search_runs([experiment.experiment_id])
    display(runs.filter(regex='run_id|params.model|metrics.test_auc_macro|metrics.test_f1_macro|metrics.test_loss'))
```
